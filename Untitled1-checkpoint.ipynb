```python
%matplotlib inline
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread(r"C:\Users\Samiran Das\OneDrive\Pictures\Screenshots\Screenshot 2024-09-07 160315.png")
img1= cv2.imread(r"C:\Users\Samiran Das\Downloads\_fb2bed02-0df4-4420-8bad-b13927671ebc.jpeg")



plt.imshow(img1)
plt.title("Original Image")
plt.show()
img.shape
plt.imshow(img)
plt.title("Original Image")
plt.show()
img.shape

```

```python
gray = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray_1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Gray Image")
plt.show()
plt.subplot(1,2,2)
plt.imshow(gray_1, cmap='gray')
plt.title("Gray Image")
plt.show()
```

```python
blur = cv2.GaussianBlur(gray, (5, 5), 0)
blut=cv2.GaussianBlur(gray_1,(5,5),0)

plt.subplot(1,2,1);
plt.imshow(blur, cmap='gray')
plt.title("Soft Blur Image")
plt.show()
plt.subplot(1,2,2);
plt.imshow(blut, cmap='gray')
plt.show()
```

```python
edges = cv2.Canny(blur, 60, 100)
plt.subplot(1,2,1)
plt.imshow(edges, cmap='gray')
plt.title("Edge Detected Image")
plt.show()
edges1 = cv2.Canny(blut, 60, 100)
plt.subplot(1,2,2)
plt.imshow(edges1, cmap='gray')
plt.show()
     
```

```python

     edges = cv2.bitwise_not(edges)
edges1 = cv2.bitwise_not(edges1)

# Save both images
cv2.imwrite("sketch1.jpg", edges)
cv2.imwrite("sketch2.jpg", edges1)

# Compute and print non-zero weights
weight1 = cv2.countNonZero(edges)
weight2 = cv2.countNonZero(edges1)
print("Sketch 1 image saved with weight:", weight1)
print("Sketch 2 image saved with weight:", weight2)

# Plot both images side-by-side
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(edges, cmap='gray')
plt.title("Pencil Sketch Image 1")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(edges1, cmap='gray')
plt.title("Pencil Sketch Image 2")
plt.axis('off')

plt.tight_layout()
plt.show()
```
